```python
def generate_intermediate_code(expr):
    """
    Generates Quadruples and Triples from a Postfix expression.
    """
    temp = 1
    quads = []
    triples = []
    stack = []

    # Standard Triple references use their index; we'll track that here
    # Tokens must be in Postfix notation for this logic to process correctly
    tokens = expr.split()

    print(f"Processing Postfix Expression: {expr}\n")

    for token in tokens:
        if token.isalnum():
            stack.append(token)
        elif token in ['+', '-', '*', '/']:
            # Pop operands
            op2 = stack.pop()
            op1 = stack.pop()

            # 1. Quadruple Generation
            temp_var = f"T{temp}"
            quads.append((token, op1, op2, temp_var))

            # 2. Triple Generation
            # We use the index of the triple for future references
            triples.append((token, op1, op2))

            # For the stack logic (to continue processing),
            # we use the temp_var for Quadruples or the index for Triples.
            # Here we follow the Quadruple path for the stack:
            stack.append(temp_var)
            temp += 1

    return quads, triples

# --- Execution ---
# Using Postfix notation: "burger fries 2 * +"
# Equivalent to: burger + (fries * 2)
postfix_expr = "burger fries 2 * +"
quadruples, triples = generate_intermediate_code(postfix_expr)

print("--- QUADRUPLES ---")
print(f"{'Op':<5} | {'Arg1':<10} | {'Arg2':<10} | {'Result'}")
print("-" * 40)
for q in quadruples:
    print(f"{q[0]:<5} | {q[1]:<10} | {q[2]:<10} | {q[3]}")

print("\n--- TRIPLES ---")
print(f"{'ID':<5} | {'Op':<5} | {'Arg1':<10} | {'Arg2'}")
print("-" * 35)
for i, t in enumerate(triples):
    print(f"({i})   | {t[0]:<5} | {t[1]:<10} | {t[2]}")
```
