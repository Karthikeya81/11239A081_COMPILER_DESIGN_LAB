```python
def count_spaces_lines(file_path):
    """
    Reads a specified text file and computes the total
    number of lines and blank spaces.
    """
    try:
        # Open the file safely using the 'with' context manager
        with open(file_path, 'r', encoding='utf-8') as f:
            text = f.read()

            # Count lines by splitting the text at newline characters
            lines = text.split('\n')
            num_lines = len(lines)

            # Count the exact number of blank space characters
            spaces = text.count(' ')

            return num_lines, spaces

    except FileNotFoundError:
        return None, "Error: The specified file was not found."

# --- Example Usage ---
# Simulating the existence of a file named 'booking.txt'
# Assume 'booking.txt' contains:
# "Room 101 Booked
#
# Customer: John Doe"

file_name = "feedback.txt"

# For the sake of this demonstration, we will print the expected execution
lines_count, spaces_count = count_spaces_lines(file_name)

if lines_count is not None:
    print(f"File Analysis for '{file_name}':")
    print("-" * 30)
    print(f"Total Lines        : {lines_count}")
    print(f"Total Blank Spaces : {spaces_count}")
else:
    print(spaces_count) # Prints the error message
```
