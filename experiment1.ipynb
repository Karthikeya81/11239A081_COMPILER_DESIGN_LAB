```python
import re

# --- Scenario 1: E-commerce Discount Rule ---
def validate_discount(expr):
    """
    Validates if a discount expression strictly consists of
    alphabetic variables separated by '+' or '-' operators.
    Parentheses or other symbols are not part of this grammar.
    """
    # Pattern: Letters, optionally followed by (+ or -) and more letters
    pattern = r'^[A-Za-z]+(\s*[\+\-]\s*[A-Za-z]+)*$'
    return bool(re.match(pattern, expr))

# Test Case for Scenario 1
expr = "A + B - (C + D)"
print(f"Expression: '{expr}'")
print(f"Grammar Validation (Discount): {validate_discount(expr)}\n")


# --- Scenario 2: Car Parking Ticket Format ---
def validate_ticket(ticket):
    """
    Validates a parking ticket against a strict format:
    Prefix 'PARK-', followed by 2+ alphanumeric chars, '-', and 4 digits.
    """
    # Pattern: PARK-[2+ Alphanumeric]-[4 Digits]
    pattern = r'^PARK\-[A-Z0-9]{2,}\-\d{4}$'
    return bool(re.match(pattern, ticket))

# Test Case for Scenario 2
ticket = "PARK-AB12-0045"
print(f"Ticket ID: '{ticket}'")
print(f"Grammar Validation (Ticket): {validate_ticket(ticket)}")
```
