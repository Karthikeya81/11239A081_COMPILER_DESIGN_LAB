```python
def shift_reduce(expr):
    """
    Performs a simple Shift-Reduce parse on an expression string.
    Specifically looks for multiplication patterns to reduce to a non-terminal.
    """
    stack = []
    # Split string into tokens (assuming space-separated input)
    tokens = expr.split()

    print(f"Initial tokens: {tokens}\n" + "-"*30)

    while tokens:
        # SHIFT: Move token from input to stack
        stack.append(tokens.pop(0))
        print(f"Shift:  {stack}")

        # REDUCE: Check if the top of the stack matches the rule: [val, '*', val]
        # This loop handles multiple reductions if applicable
        while len(stack) >= 3 and stack[-2] == '*':
            # Replace the top 3 elements with the non-terminal 'E'
            stack[-3:] = ['E']
            print(f"Reduce: {stack}")

    return stack

# --- Test Execution ---
expression = "price + discount * quantity"
final_stack = shift_reduce(expression)

print("-"*30)
print(f"Final Stack State: {final_stack}")
```
