```python
class SymbolTable:
    """
    A class-based implementation of a Symbol Table
    typically used by compilers to manage identifiers.
    """
    def __init__(self):
        # Using a dictionary for constant time O(1) average lookup
        self.table = {}

    def insert(self, key, value):
        """Adds or updates a symbol in the table."""
        self.table[key] = value

    def lookup(self, key):
        """Retrieves information for a given symbol."""
        return self.table.get(key, "Not Found")

# --- Execution and Testing ---

# Initialize the Symbol Table
sym = SymbolTable()

# Insert an identifier with associated metadata (attributes)
identifier = "Cake"
metadata = {"price": 50, "stock": 20}
sym.insert(identifier, metadata)

# Perform Lookup
print(f"Searching for identifier: '{identifier}'")
result = sym.lookup(identifier)
print(f"Result: {result}")

# Testing an unknown identifier
print(f"\nSearching for identifier: 'Bread'")
print(f"Result: {sym.lookup('Bread')}")
```
