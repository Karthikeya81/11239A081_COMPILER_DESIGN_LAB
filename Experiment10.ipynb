```python
def tokenize(expr):
    """
    Simulates the tokenization phase of a compiler.
    Categorizes lexemes into Identifiers (ID) and Operators (OP).
    """
    tokens = []

    # Splitting the expression based on whitespace
    lexemes = expr.split()

    for t in lexemes:
        # Check if the lexeme consists of alphabetic characters
        if t.isalpha():
            tokens.append(("ID", t))
        else:
            # Assume non-alphabetic lexemes are operators in this simple model
            tokens.append(("OP", t))

    return tokens

# --- Test Case ---
source_code = "ITEM + ITEM - DISCOUNT"
generated_tokens = tokenize(source_code)

print(f"Source Expression: {source_code}")
print(f"Generated Tokens:  {generated_tokens}")
```
