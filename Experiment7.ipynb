```python
def count_chars(s):
    """
    Analyzes the input string and counts the occurrences of
    alphabets, digits, whitespaces, and other characters.
    """
    # Initialize counters
    letters = digits = spaces = others = 0

    # Iterate through each character in the string
    for ch in s:
        if ch.isalpha():
            letters += 1
        elif ch.isdigit():
            digits += 1
        elif ch.isspace():
            spaces += 1
        else:
            # Captures punctuation, symbols, and special characters
            others += 1

    return letters, digits, spaces, others

# --- Test Cases ---

# Case 1: Mixed string with letters, numbers, spaces, and symbols
test_str1 = "Pizza 123! #1"
l1, d1, s1, o1 = count_chars(test_str1)
print(f"Input: '{test_str1}'")
print(f"Letters: {l1}, Digits: {d1}, Spaces: {s1}, Others: {o1}\n")

# Case 2: Alphanumeric string with no spaces or symbols
test_str2 = "MH12AB1234"
l2, d2, s2, o2 = count_chars(test_str2)
print(f"Input: '{test_str2}'")
print(f"Letters: {l2}, Digits: {d2}, Spaces: {s2}, Others: {o2}")
```
