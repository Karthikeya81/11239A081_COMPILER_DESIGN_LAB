```python
def infix_to_postfix(expr):
    """Standard Shunting-yard algorithm to convert infix to postfix."""
    prec = {'+': 1, '-': 1, '*': 2, '/': 2, '(': 0}
    stack = []
    output = []

    for token in expr.split():
        if token.isalnum():
            output.append(token)
        elif token == '(':
            stack.append(token)
        elif token == ')':
            while stack and stack[-1] != '(':
                output.append(stack.pop())
            stack.pop() # Remove '('
        else:
            while stack and prec.get(stack[-1], 0) >= prec.get(token, 0):
                output.append(stack.pop())
            stack.append(token)

    while stack:
        output.append(stack.pop())
    return " ".join(output)

def reverse_expr(s):
    """
    Reverses the tokens of an expression and swaps parentheses.
    Reversing tokens as units (e.g., 'price') is essential for prefix notation.
    """
    tokens = s.split()[::-1]
    for i in range(len(tokens)):
        if tokens[i] == '(':
            tokens[i] = ')'
        elif tokens[i] == ')':
            tokens[i] = '('
    return " ".join(tokens)

def infix_to_prefix(expr):
    """Converts Infix to Prefix using the reverse-postfix-reverse method."""
    # 1. Reverse the infix expression
    rev = reverse_expr(expr)
    # 2. Convert reversed expression to postfix
    postfix = infix_to_postfix(rev)
    # 3. Reverse the postfix result to get prefix
    prefix = " ".join(postfix.split()[::-1])
    return prefix

# --- Execution ---
input_expr = "price + discount * quantity"
result = infix_to_prefix(input_expr)

print(f"Infix Expression  : {input_expr}")
print(f"Prefix Expression : {result}")
```
