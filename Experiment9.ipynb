```python
def wc(filename):
    """
    Reads a file and returns the count of lines, words, and characters.
    """
    try:
        # Opening the file using 'with' ensures proper resource management
        with open(filename, 'r', encoding='utf-8') as f:
            text = f.read()

            # Line count: Counts newlines and adds 1 for the last line (if file not empty)
            lines = text.count('\n') + 1 if text else 0

            # Word count: Splits by whitespace and counts the elements
            words = len(text.split())

            # Character count: Total length of the string
            chars = len(text)

            return lines, words, chars

    except FileNotFoundError:
        return "Error: File not found."

# --- Example Usage ---
# Assuming 'feedback.txt' contains:
# "Hello World
# This is a lab report."

file_name = "feedback.txt"
# For demonstration, we simulate the output based on sample text
result = wc(file_name)

if isinstance(result, tuple):
    l, w, c = result
    print(f"File Statistics for '{file_name}':")
    print(f"{'Lines':<12}: {l}")
    print(f"{'Words':<12}: {w}")
    print(f"{'Characters':<12}: {c}")
else:
    print(result)
```
