```python
%%writefile my_lexer.py
import ply.lex as lex

# 1. List of token names. This is always required.
tokens = [
    'ADD',
    'ID',
    'NUMBER'
]

# 2. Regular expression rules for simple tokens
t_ADD    = r'ADD'
t_ID     = r'[a-zA-Z_][a-zA-Z0-9_]*'
t_NUMBER = r'\d+'

# 3. A string containing ignored characters (spaces and tabs)
t_ignore = ' \t'

# 4. Define a rule so we can track line numbers
def t_newline(t):
    r'\n+'
    t.lexer.lineno += len(t.value)

# 5. Error handling rule
def t_error(t):
    print(f"Illegal character '{t.value[0]}'")
    t.lexer.skip(1)

# Build the lexer
lexer = lex.lex()

if __name__ == "__main__":
    # Test input
    data = "ADD item1 2"

    # Give the lexer some input
    lexer.input(data)

    # Tokenize and print results
    print(f"Input string: '{data}'")
    print("-" * 30)
    for tok in lexer:
        print(tok)
```
