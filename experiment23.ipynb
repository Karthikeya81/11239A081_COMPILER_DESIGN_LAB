```python
def infix_to_postfix(expr):
    """
    Converts an infix string expression to postfix notation
    using the Shunting-yard algorithm.
    """
    # Define precedence levels for operators
    prec = {'+': 1, '-': 1, '*': 2, '/': 2, '(': 0}
    stack = []
    output = []

    # Split the expression into tokens (assuming space-separated input)
    tokens = expr.split()

    for token in tokens:
        # If the token is an operand, add it to the output
        if token.isalpha() or token.isdigit():
            output.append(token)

        # If the token is '(', push it onto the operator stack
        elif token == '(':
            stack.append(token)

        # If the token is ')', pop from stack until '(' is found
        elif token == ')':
            while stack and stack[-1] != '(':
                output.append(stack.pop())
            # Remove the '(' from the stack
            stack.pop()

        # If the token is an operator
        else:
            # Pop operators with higher or equal precedence
            while stack and prec.get(stack[-1], 0) >= prec.get(token, 0):
                output.append(stack.pop())
            # Push current operator to stack
            stack.append(token)

    # Append any remaining operators from the stack to the output
    while stack:
        output.append(stack.pop())

    return " ".join(output)

# --- Test Execution ---
infix_expression = "price + discount * quantity"
postfix_result = infix_to_postfix(infix_expression)

print(f"Infix Expression   : {infix_expression}")
print(f"Postfix Expression : {postfix_result}")
```
