```python
def generate_code(tac):
    """
    Translates Three-Address Code (TAC) quadruples into
    human-readable code statements.
    """
    generated_statements = []

    for op, arg1, arg2, result in tac:
        if op == '=':
            # Simple assignment: result = arg1
            statement = f"{result} = {arg1}"
        else:
            # Binary operation: result = arg1 op arg2
            statement = f"{result} = {arg1} {op} {arg2}"

        generated_statements.append(statement)

    return generated_statements

# --- Execution ---

# Input TAC representing: total = a * (b + c)
# Quadruple format: (op, arg1, arg2, result)
three_addr_code = [
    ('+', 'b', 'c', 't1'),
    ('*', 'a', 't1', 't2'),
    ('=', 't2', '', 'total')
]

print("Input Three-Address Code (Quadruples):")
for instr in three_addr_code:
    print(instr)

print("\n--- Generated Code ---")
code_output = generate_code(three_addr_code)
for line in code_output:
    print(line)
```
