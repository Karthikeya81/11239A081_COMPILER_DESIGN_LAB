```python
%%writefile vowel_lexer.py
import ply.lex as lex

# 1. Define the tokens
tokens = ['VOWEL', 'CONSONANT']

# 2. Rule for Vowels
def t_VOWEL(t):
    r'[aeiouAEIOU]'
    return t

# 3. Rule for Consonants (matches any letter not caught by t_VOWEL)
def t_CONSONANT(t):
    r'[a-zA-Z]'
    return t

# 4. Ignored characters
t_ignore = ' \t\n'

# 5. Error handling
def t_error(t):
    print(f"Illegal character '{t.value[0]}'")
    t.lexer.skip(1)

# Build the lexer
lexer = lex.lex()

if __name__ == "__main__":
    # Test Input
    data = "Bakery"
    lexer.input(data)

    print(f"Analyzing input: '{data}'")
    print("-" * 30)

    # Tokenize
    for tok in lexer:
        # Printing the standard LexToken representation
        print(tok)
```
