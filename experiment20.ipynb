```python
%%writefile char_lexer.py
import ply.lex as lex

# 1. Define the token names
tokens = ['CHAR']

# 2. Define the rule for non-whitespace characters
def t_CHAR(t):
    r'[^\s]'
    return t

# 3. Define the whitespace characters to be ignored
t_ignore = ' \t\n'

# 4. Error handling for illegal characters
def t_error(t):
    print(f"Illegal character '{t.value[0]}'")
    t.lexer.skip(1)

# Build the lexer
lexer = lex.lex()

if __name__ == "__main__":
    # Test input
    input_data = "This is a test"
    lexer.input(input_data)

    print(f"Original Input: '{input_data}'")
    print("Processed Output: ", end="")

    # 5. Tokenize and print only the non-whitespace characters
    for tok in lexer:
        print(tok.value, end="")
    print() # Final newline for console formatting
```
