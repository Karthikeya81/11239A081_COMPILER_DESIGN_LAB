```python
def shift_reduce(expr):
    """
    Simulates a Shift-Reduce parser.
    It 'shifts' tokens onto a stack and 'reduces' them to a non-terminal 'E'
    whenever a multiplication pattern is detected at the top of the stack.
    """
    stack = []
    # Tokenizing the expression by whitespace
    tokens = expr.split()

    print(f"Input Tokens: {tokens}\n" + "="*40)

    while tokens:
        # SHIFT: Move the next token from input to the stack
        current_token = tokens.pop(0)
        stack.append(current_token)
        print(f"Shift  : {stack}")

        # REDUCE: Check for the handle [Any, '*', Any] at the end of the stack
        # This while loop allows for recursive reductions if necessary
        while len(stack) >= 3 and stack[-2] == '*':
            # Reduce the three-token sequence to the non-terminal 'E'
            stack[-3:] = ['E']
            print(f"Reduce : {stack}")

    return stack

# --- Test Execution ---
# Note: Ensure spaces are present for proper tokenization via split()
input_expr = "principal + (principal * rate)"
result_stack = shift_reduce(input_expr)

print("="*40)
print(f"Final Stack State: {result_stack}")
```
