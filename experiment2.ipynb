```python
def is_balanced(expr):
    """
    Checks if the given expression has balanced parentheses using a stack.
    """
    stack = []

    # Traverse each character in the expression
    for ch in expr:
        if ch == '(':
            # Push opening parenthesis onto the stack
            stack.append(ch)
        elif ch == ')':
            # If stack is empty, we have a closing parenthesis without an opening one
            if not stack:
                return False
            # Pop the matching opening parenthesis
            stack.pop()

    # If the stack is empty, all parentheses matched. Otherwise, unbalanced.
    return len(stack) == 0

# --- Test Cases ---

# Case 1: Unbalanced expression (extra closing parenthesis)
expr1 = "(deposit + withdraw - transfer))"
print(f"Expression 1: '{expr1}'")
print(f"Is Balanced : {is_balanced(expr1)}\n")

# Case 2: Balanced expression
expr2 = "((math + science) + (english + social))"
print(f"Expression 2: '{expr2}'")
print(f"Is Balanced : {is_balanced(expr2)}")
```
