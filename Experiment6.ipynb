```python
# Lexical Analyzer Simulation: Operator Validation

def validate_op(expr):
    """
    Validates the presence of relational multi-character operators.
    """
    # Define the set of operators to recognize
    operators = ["<=", ">=", "==", "!="]

    # Check if any operator in the list exists within the expression
    return any(op in expr for op in operators)

def check_speed(expr):
    """
    Simulates detection of increment (++) and decrement (--) operators.
    Note: In a formal lexical analyzer, further checks would be needed
    to ensure strict grammar (e.g., handling '+++' as an error).
    """
    return "++" in expr or "--" in expr

# --- Test Executions ---

# Testing Relational Operators
expr1 = "balance >= 5000"
print(f"Expression: '{expr1}' | Contains valid operator: {validate_op(expr1)}")

# Testing Unary Operators
expr2 = "speed++"
print(f"Expression: '{expr2}' | Contains increment/decrement: {check_speed(expr2)}")

# Testing Edge Case
expr3 = "speed+++"
print(f"Expression: '{expr3}' | Contains increment/decrement: {check_speed(expr3)}")
```
