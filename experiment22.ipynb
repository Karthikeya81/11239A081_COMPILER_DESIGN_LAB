```python
def eval_postfix(expr):
    """
    Evaluates a Postfix expression using a stack-based approach.
    Expects a space-separated string of operands and operators.
    """
    stack = []

    # Process each token in the input string
    for token in expr.split():
        # Check if the token is a numeric operand
        if token.isdigit():
            stack.append(int(token))
        else:
            # Token is an operator; pop operands from the stack
            # Note: b is the top element (second operand), a is the one below (first)
            b = stack.pop()
            a = stack.pop()

            # Perform the relevant arithmetic operation
            if token == '+':
                stack.append(a + b)
            elif token == '-':
                stack.append(a - b)
            elif token == '*':
                stack.append(a * b)
            elif token == '/':
                # Use floor division to maintain integer results
                stack.append(a // b)

    # The final result remains at the top of the stack
    return stack[0]

# --- Test Case ---
postfix_input = "2 3 + 5 *"
result = eval_postfix(postfix_input)

print(f"Postfix Expression : {postfix_input}")
print(f"Evaluated Result   : {result}")
```
