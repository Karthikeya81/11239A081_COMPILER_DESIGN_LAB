```python
def is_keyword(word, keywords):
    """
    Checks if a given word exists within a specified set of keywords.
    Converts the input to uppercase to guarantee case-insensitive comparison.
    """
    # Using 'in' with a set provides an efficient O(1) average time complexity lookup
    return word.upper() in keywords

# --- Scenario 1: E-commerce Actions ---
keywords1 = {"ORDER", "CANCEL", "TRACK"}
test_word1 = "TRACK"

print(f"--- E-commerce Context ---")
print(f"Keywords: {keywords1}")
print(f"Is '{test_word1}' a valid keyword? {is_keyword(test_word1, keywords1)}\n")

# --- Scenario 2: Inventory/Library Actions ---
keywords2 = {"ISSUE", "RETURN", "SEARCH"}
test_word2 = "RETURN"
test_word3 = "BORROW" # Negative test case

print(f"--- Inventory Context ---")
print(f"Keywords: {keywords2}")
print(f"Is '{test_word2}' a valid keyword? {is_keyword(test_word2, keywords2)}")
print(f"Is '{test_word3}' a valid keyword? {is_keyword(test_word3, keywords2)}")
```
